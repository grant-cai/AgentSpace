```python
from Agent import personAgent
import json
```

```python
agent = personAgent("essay_chunk_agentspace", "chunk_index")
```

```python
# Load evaluation questions
with open("../Evaluation/test_questions.json", "r") as f:
    questions = json.load(f)
```

```python
# Run evaluation loop for the first few questions
for item in questions[:3]:
    thread_id = agent.new_thread_id()
    question = item['question']
    print(f"--- Question ---\n{question}\n")
    
    response = agent.chat(question, thread_id)
    agent_response = response[0]['text'] if response else "No response"
    
    print(f"--- Agent Response ---\n{agent_response}\n")
    print(f"--- Ground Truth ---\n{item['ground_truth']}\n")
    print("=" * 80 + "\n")
```
